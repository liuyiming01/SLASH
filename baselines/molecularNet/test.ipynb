```python
from rdkit import Chem

def smiles_to_custom_graph(smiles):
    # 1. 将 SMILES 转化为 RDKit 分子对象
    mol = Chem.MolFromSmiles(smiles)
    if not mol:
        return "Invalid SMILES"

    # 2. 提取节点 (原子) 信息
    # 格式: [index, weight]
    # 使用原子序数 (AtomicNum) 作为权重
    node_weights = []
    atom_details = [] # 用于辅助理解，打印出原子符号
    
    for atom in mol.GetAtoms():
        idx = atom.GetIdx()
        weight = atom.GetAtomicNum() # 碳=6, 氧=8, 等
        symbol = atom.GetSymbol()
        
        node_weights.append(f"[{idx}, {weight}]")
        atom_details.append(f"{idx}:{symbol}")

    # 3. 提取边 (化学键) 信息
    # 格式: (u, v)
    # RDKit 的键是无向的，但通常列出一次即可
    edges = []
    edge_weights = [] # 如果你需要边的权重
    
    for bond in mol.GetBonds():
        u = bond.GetBeginAtomIdx()
        v = bond.GetEndAtomIdx()
        
        # 简单的 (u, v) 格式
        edges.append(f"({u}, {v})")
        
        # 进阶: 提取键级作为边的权重
        # 1.0=单键, 2.0=双键, 1.5=芳香键
        b_type = bond.GetBondTypeAsDouble()
        edge_weights.append(f"({u}, {v}, {b_type})")

    # 4. 组装输出文本
    output_text = (
        f"The nodes are numbered from 0 to {mol.GetNumAtoms() - 1}.\n"
        f"Weights of nodes (Atomic Number) are: {' '.join(node_weights)}\n"
        f"The edges are: {' '.join(edges)}"
    )
    
    return output_text, atom_details, edge_weights

# 你的输入分子
input_smiles = "Cc1onc(c2ccccc2Cl)c1C(=O)N[C@H]3[C@H]4SC(C)(C)[C@@H](N4C3=O)C(O)=O"

graph_text, atoms, weighted_edges = smiles_to_custom_graph(input_smiles)

print("--- 适配你当前方法的序列化输出 ---")
print(graph_text)

print("\n--- 辅助调试信息 (节点对应原子) ---")
print(atoms)

print("\n--- 如果你的方法支持边权重 (Bond Order) ---")
print(' '.join(weighted_edges))
```
